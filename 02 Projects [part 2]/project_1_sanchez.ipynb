##### This project examines country-year transparency scores using the 2024 [HRV Transparency Dataset](https://hrvtransparency.org/download-data/). We focus on the `transparencyindex` variable, calculating its mean, meadian and mode in two ways before visualizing the average score by year. 

```python
import pandas as pd
import numpy as np

hrv = pd.read_csv("TransparencyData.csv")
hrv.head()

hrv_1 = hrv[["Country", "year", "transparencyindex"]]
hrv_1.head()
```

```python
# mean, median and mode
hrv_mean = hrv_1["transparencyindex"].mean()
hrv_median = hrv_1["transparencyindex"].median()
hrv_mode = hrv_1["transparencyindex"].mode()

print("Mean:", hrv_mean)
print("Median:", hrv_median)
print("Mode(s):", hrv_mode)
```

```python
import csv

# Load the scores without pandas.
scores = []

with open("TransparencyData.csv", newline="") as file:
    for row in csv.DictReader(file):
        scores.append(float(row["transparencyindex"]))

# Mean
mean_score = sum(scores) / len(scores)

# Median
scores.sort()
n = len(scores)
middle = n // 2

if n % 2 == 1:
    median_score = scores[middle]
else:
    median_score = (scores[middle - 1] + scores[middle]) / 2

# Count how often each score appears.
counts = {}

for score in scores:
    if score in counts:
        counts[score] += 1
    else:
        counts[score] = 1

# Mode: keep every score tied for the highest count.
highest_count = max(counts.values())
modes = []

for score in counts:
    if counts[score] == highest_count:
        modes.append(score)

print("Mean:", mean_score)
print("Median:", median_score)
print("Mode(s):", modes)
```

**The mean transparency score is approximately 1.840, and the median is approximately 1.108. Both calculation methods give the same results. Every score occurs once, so there is no unique mode.**

#### Visualizing Government Transparency Over Time

```python
# Calculate the average transparency index for each year.
yearly_means = hrv_1.groupby("year")["transparencyindex"].mean().dropna()

# Scale the bars to a maximum of 30 stars.
largest = yearly_means.abs().max()
scale = 30 / largest if largest != 0 else 1

print("Average transparency index by year")
print("Bars extend left for negative values and right for positive values.")
print(f"Each * represents approximately {1 / scale:.3f} index units.\n")

for year, value in yearly_means.items():
    bar = "*" * round(abs(value) * scale)
    if value < 0:
        print(f"{year}: {bar:>30}| {value:.3f}")
    else:
        print(f"{year}: {'':>30}|{bar} {value:.3f}")
```

**The average transparency index generally rises from 1980 to the early 2010s, then declines through 2015. These are averages across the available country observations each year.**
